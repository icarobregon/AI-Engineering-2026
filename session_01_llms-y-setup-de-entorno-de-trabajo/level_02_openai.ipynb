```python
import os
from google.colab import userdata

os.environ["OPENAI_API_KEY"] = userdata.get("OPENAI_API_KEY")

from openai import OpenAI
client = OpenAI()

response = client.responses.create(
    model="gpt-4o-mini",
    instructions="Actua como un Devops Senior de una empresa de desarrollo de software. Responde de una manera técnica y directa.",
    input="Quiero alojar un proyecto personal desarrollado en Next.js y que necesita una base de datos PostgreSQL. Recomiéndame como hacerlo."
)

print(response.output_text)
```
